```python
def print_date(year,month,day):
    joined = str(year) + '/' + str(month) + '/' + str(day)
    return joined

print(print_date(month=9, year = 2026, day=13)
```

```python
def average(values):
    if len(values) == 0:
        return
    return sum(values)/len(values)

print(average([2,4,6,8]))
```
